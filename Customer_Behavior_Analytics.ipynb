Now, Replaced the missing with mean/ median, mean can be affected by outliers skewing the data distribution, median is robust to outliers hence median is chosen. Overall median replacement could be biased as reviews are offered for various items/ category. So Impute the median review rating within each category separate. 

```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
# feature engineering, create a column for age_group
labels = ['Youth', 'Adult', 'Mature', 'Senior']
df['age_group'] = pd.qcut(df['age'], q = 4, labels = labels)
```

- pd.qcut() divides a column into equal-sized groups (quantiles)

q = 4 means:
→ split the age column into 4 groups, each containing 25% of the data.

So instead of dividing by fixed age ranges (like 18–25), it divides based on the distribution of your data.

```python
# age_group is a new column showing which age quantile each person belongs to
df[['age', 'age_group']].head(10)
```

```python
# creating purchase_frequency_days column
# textual frequencies to number for quicker analysis
frequency_mapping = {
    'Every 3 Months':    90,
'Annually'    :    365,
'Quarterly'     :   90,
'Monthly'     :  30,
'Bi-Weekly'      : 14,
'Fortnightly'      :14,
'Weekly'      :     7 }
df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```
